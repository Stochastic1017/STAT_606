Let's recall our classes `Point` and `Rectangle`, defined in lecture.
Here are slightly different implementations that make use of attributes and a couple of methods.

```python
class Point:
    
    def __init__( self, x, y ):
        # Ideally, we would include some error checking here,
        # to verify that x and y are both numeric (ints or floats).
        # This is just demo code, so we're skipping that,
        # but if you're bored, try adding some error checking here.
        self.x = x
        self.y = y
    
    # These "get" methods are a common convention in object oriented programming.
    # The idea is that your end-user should never have to "reach into" the object
    # to do things like point.x = 4.
    # Instead, we want a user of our object to only ever interact with the object via methods.
    def get_x(self):
        return self.x
    def get_y(self):
        return self.y
    
    # If this were "real" code, we would also want these "set" methods,
    # which I am only including for the sake of example.
    def set_x(self, x):
        self.x = x
    def set_y(self, y):
        self.y = y
```

```python
class Rectangle:
    def __init__( self, corner, w, h ):
        # Ideally, we would include some error checking here,
        # to verify that corner is a Point and that w and h are non-negative numerics.
        # This is just demo code, so we're skipping that,
        # but if you're bored, try adding some error checking here.
        self.corner = corner
        self.width = w
        self.height = h
        
    def get_corner( self ):
        return self.corner
    
    def get_width( self ):
        return self.width
    
    def get_height( self ):
        return self.height
```

<b>Hint:</b> I've written a function `squared_distance` that you may find useful: it takes two `Point` objects and returns the square of the distance between those two points. Be sure to read through it carefully and convince yourself that it does what it claims.

```python
def squared_dist( p1, p2 ):
    dxsquared = (p1.get_x()-p2.get_x())**2
    dysquared = (p1.get_y()-p2.get_y())**2
    return dxsquared + dysquared
    
def point_in_circle( circ, pt ):
    # Compare squared distance between pt and circ.center
    # to squared circ.radius
    
    sqdist = squared_dist( pt, circ.get_center() )
    
    return sqdist <= circ.get_radius()**2
```

```python
def what_day_is_it():
    # Get the current datetime
    d = datetime.datetime.now()
    
    # Extract the isoweekday from that datetime object
    isowd = d.isoweekday()
    
    # Translate that isoweekday, which is a number,
    # into a string representing the day of the week
    # (e.g., 'Monday', 'Tuesday', etc.).
    daylist = ['','Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
    
    # Print that day of the week.
    print( daylist[isowd] )
```

```python
what_day_is_it() # Should return Monday, if you're running this on Monday, February 19, 2024
```
