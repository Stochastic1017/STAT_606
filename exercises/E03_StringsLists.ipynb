```python
def char_count( s, c ):
    # c should be a character (i.e., a string of length 1)
    if not isinstance( s, str):
        raise TypeError('s should be of type str')
    if not isinstance( c, str):
        raise TypeError('c should be of type str')
    # At this point we know that c must be a string
    # but we don't yet know that it has length 1.
    if len(c) != 1:
        raise ValueError('c should be a character, i.e., a string of length 1')
        
    return s.count(c)
```

```python
def rotate_char( c, r ):
    #TODO: error checking
    # Assume for now that c is a length-1 string and r is an integer
    
    # Instead of doing math on the integers 113 to 139ish, let's just turn all
    # characters into numbers 0 to 25
    n = ord(c) - ord('a') # n is encoding of c as its position in alphabet.
    rotated_n = (n + r) % 26 # rotate around the alphabet
    rotated_ord = ord('a') + rotated_n # rotated_ord should be in [113:ord('z')]
    assert( rotated_ord >= ord('a') and rotated_ord <= ord('z') )
    return chr( rotated_ord )

def rotate_word( s, r ):
    #TODO: error checking
    #Assume for now that s is a string of all lower-case and r is an integer.

    # Hint: you might find it useful to write a "helper function" that rotates an individual character
    # by a certain amount. Something like rotate_char( c, r ).
    # Let's pretend for now that we have such a function.
    result = ''
    for c in s:
        result = result + rotate_char( c, r )
    
    return result
```

## Problem 5: Duplicates

Write a function called `has_duplicates` that takes a list as its only argument and returns `True` if there is any element that appears more than once and `False` otherwise. It should not modify the original list.

```python
# Note that this is not the most efficient solution to this problem.
# We'll see a better (i.e., faster) solution when we talk about dictionaries.
def has_duplicates( t ):
    # TODO: error checking 

    # We are going to traverse the list, and for each element
    # that we find, we'll look if it has matches in the rest
    # of the list.
    for i in range( len(t) ):
        current_elmt = t[i]
        for j in range( i+1, len(t)):
            if current_elmt == t[j]:
                return True # We found a duplicate
    # If we got to the end of the for-loop, we must have
    # never found any duplicates, so return False.
    False
```

```python
# Here's an example of "unit testing"-- this is more or less what the grading script does!
assert( has_duplicates([1,2,3,4,5,1]) ) # Remember, assert(expr) raises an error if expr is NOT true.
assert( not has_duplicates([1,2,3,4,5]) )
assert( not has_duplicates([]) ) # always important to check weird edge cases like the empty list!
assert( has_duplicates([1,2,2,3]))
assert( has_duplicates([2,2,3,4,5,5]))
assert( has_duplicates([1,2,3,4,1]))
```
