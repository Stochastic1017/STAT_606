## Problem 2: Basic Indexing

The block of code below creates a table containing some basic information about the colleges and universities with the most NCAA Division I Ice Hockey championships (see <a href="https://en.wikipedia.org/wiki/NCAA_Division_I_Men's_Ice_Hockey_Tournament">here</a> for more information).

```python
schools = ['Michigan', 'Denver', 'North Dakota', 'Wisconsin', 'Boston College',
           'Boston University', 'Minnesota', 'Lake Superior State']
titles = [9,9,8,6,5,5,5,3]
states = ['MI', 'CO', 'ND', 'WI', 'MA', 'MA', 'MI', 'MI']
hockey_dict = {'College':pd.Series(schools), '#Titles':pd.Series(titles), 'State':pd.Series(states) }
h = pd.DataFrame( hockey_dict )
h
```

### 2.1
Oops! There's a typo in there-- U. Minnesota is in Minnesota (state abbreviation MN), not Michigan!
Correct the typo (using indexing, not by modifying the code above!).

```python
h['College']
```

```python
h['#Titles']
```

```python
# Here's what we would LIKE to do, but pandas yells at us.
# Note that this is just a warning; this works fine
h['State'][6] = 'MN'
```

```python
# Note that if the above series were on its own, rather than in a DataFrame, this would be fine.
state_series = pd.Series(states)
state_series
```

```python
# Reset the entry to be "wrong"
h['State'][6] = 'MI'
```

```python
h.iat[6,2] = 'MN' # Also works; pick our row and column via their integer indices.
```

Modify the code above so that the rows are indexed by the schools' names, instead of by the integers 0 to 7.

```python
# Here's the "obvious" way to do it, but we see that this runs into trouble
schools = ['Michigan', 'Denver', 'North Dakota', 'Wisconsin', 'Boston College',
           'Boston University', 'Minnesota', 'Lake Superior State']
titles = [9,9,8,6,5,5,5,3]
states = ['MI', 'CO', 'ND', 'WI', 'MA', 'MA', 'MI', 'MI']
hockey_dict = {'College':pd.Series(schools), '#Titles':pd.Series(titles), 'State':pd.Series(states) }
h = pd.DataFrame( hockey_dict, index=schools )
h
```

Add a column called `InWisconsin?` to this `DataFrame`, whose values are Booleans encoding whether or not each of these universities is in Wisconsin.

```python
schools = pd.Series( ['Michigan', 'Denver', 'North Dakota', 'Wisconsin', 'Boston College',
           'Boston University', 'Minnesota', 'Lake Superior State'] )
titles = pd.Series( [9,9,8,6,5,5,5,3] )
states = pd.Series( ['MI', 'CO', 'ND', 'WI', 'MA', 'MA', 'MI', 'MI'] )
hockey_dict = {'College':schools, '#Titles':titles, 'State':states }
h = pd.DataFrame( hockey_dict )
h
```

Hmm. What's up with that extra column?
Well, `read_csv` expects certain behaviors by default.
In particular, it doesn't know whether or not the column of index labels in the `csv` file is supposed to encode a column of data or the index labels. We have to tell this `to pandas`! Refer to the documentation on the `index_col` keyword for more information.

```python
h2 = pd.read_csv('ncaa_hockey.csv', index_col=0) 
h2
```
