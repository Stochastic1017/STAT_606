```python
import math
def my_sqrt(x):
    try:
        return math.sqrt(x)
    except TypeError:
        print('Type does not allow sqrt.')
    except ValueError:
        print('Negative?')
    except:
        print('Gosh, no idea what happened.')
```

```python
import math

def is_prime(n):
    if n <= 1:
        return False
    elif n==2:
        return True
    else:
        ulim = math.ceil(math.sqrt(n))
        for k in range(2,ulim+1):
            if n%k==0:
                return False
        return True
def is_square(n):
    r = int(math.sqrt(n))
    return(r*r==n or (r+1)*(r+1)==n)
```
