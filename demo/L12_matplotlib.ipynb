```python
K=12; n=25
draws = np.zeros((n,K))
for k in range(K):
    mu = np.sin(2*np.pi*k/K)
    draws[:,k] = np.random.normal(mu,1,n)
_ = plt.boxplot(draws, labels=list('JFMAMJJASOND'))
```

```python
import scipy.stats
```

```python
mu=np.array([0.5,0.5])
Sigma=np.array([[0.1,0.05],[0.05,0.1]])
mvn1 = scipy.stats.multivariate_normal(mu,Sigma)

x, y = np.mgrid[0:1:.01, 0:1:.01]
pos = np.empty(x.shape + (2,))
pos[:, :, 0] = x; pos[:, :, 1] = y

_ = plt.contour(x, y, mvn1.pdf(pos))
```

```python
t = np.arange(20)+1
plt.subplot(221)
plt.plot(t,t,'-+b')
plt.title('linear')
plt.subplot(222)
plt.plot(t,t**2,':^r')
plt.title('quadratic')
plt.subplot(223)
plt.plot(t,np.sqrt(t),'--sg')
plt.title('sqrt')
plt.subplot(224)
plt.plot(t,np.log(t),'-oy')
plt.title('logarithmic')
```
