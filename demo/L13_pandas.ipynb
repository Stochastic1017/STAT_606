# Data Transformations

```python
index = pd.date_range('10/1/1999', periods=1100)
ts = pd.Series(np.random.normal(0.5, 2, 1100), index)
ts.head()
```

```python
key = lambda d: d.year
zscore = lambda x: (x - x.mean()) / x.std()
transformed = ts.groupby(key).transform(zscore)
transformed.head()
```

```python
transformed.describe()
```

# Pivoting and Stacking

```python
tuples = list(zip(*[['bird','bird','goat','goat'],['x','y','x','y']]))
index = pd.MultiIndex.from_tuples( tuples, names=['animal','cond'])
df = pd.DataFrame(np.random.randn(4,2), index=index, columns=['A','B'])
df
```

```python
df.stack()
```

```python
s = df.stack()
s['bird']['x']['A']
```

```python
columns = pd.MultiIndex.from_tuples([('A','cat','long'),('B','cat','long'),
                                    ('A','dog','short'),('B','dog','short')],
                                    names=['cond','animal','hair_length'])
df = pd.DataFrame( np.random.randn( 4,4 ), columns=columns )
df
```

```python
df.stack(level=1)
```

```python
df.stack(level=[0,1,2])
```
