```python
import math

def is_prime(n):
    if n <= 1:
        return False
    elif n==2:
        return True
    else:
        ulim = math.ceil(math.sqrt(n))
        for k in range(2,ulim+1):
            if n%k==0:
                return False
        return True

# NOTE: running this code in Jupyter will cause an error--
# Jupyter is running in namespace __main__, but there are no command-line args!
#if __name__ == "__main__":
#    import sys
#    USAGE='python ' + sys.argv[0] + ' int'
#    if len(sys.argv) != 2:
#        raise RuntimeError(USAGE)
#    else:
#        n = int( sys.argv[1] )
#        if is_prime(n):
#            print(sys.argv[1]+' is prime.')
#        else:
#            print(sys.argv[1]+' is not prime.')
```
