```python
x = np.arange( 1, 11 )
for _ in range(5):
    # x/float(sum(x)) is a vector that sums to 1; specifies the probability of choosing each elmt of x
    print( np.random.choice(x, 5, False, x/float(sum(x)) ) )
```

```python
x = np.random.normal(0, 1, 100) # Generate 100 draws from a normal
np.mean(x), np.median(x), np.std(x)
```

```python
np.min(x), np.max(x), np.ptp(x) # ptp is max-min
```

```python
np.std(x), np.var(x)
```

```python
x[5] = np.nan # Set 5-elmt of x to be NaN ("not a number")
np.mean(x) # and now the mean fails-- operation on a nan results in another nan.
```

```python
np.nanmin(x), np.nanmax(x), np.nanstd(x), np.nanvar(x) # Tell numpy to drop NaNs.
```
