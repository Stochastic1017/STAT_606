**Write a function $\texttt{reverse}\_\texttt{lines}$, called as $\texttt{reverse}\_\texttt{lines(infile, outfile)}$, where $\texttt{infile}$ and $\texttt{outfile}$ are strings specifying file names. Calling $\texttt{reverse}\_\texttt{lines(infile, outfile)}$ should overwrite $\texttt{outfile}$ so that it is a copy of $\texttt{infile}$, except that each line of $\texttt{infile}$ is reversed. So, for example, if the file $\texttt{infile}$ has the contents**

$\texttt{hello}$

$\texttt{tacocat}$

$\texttt{1 2 3 4 5}$

**then the file $\texttt{outfile}$ should read**

$\texttt{olleh}$

$\texttt{tacocat}$

$\texttt{5 4 3 2 1}$

**You may assume that every line of $\texttt{infile}$ ends with a newline. Likewise, every line of $\texttt{outfile}$ should end with a newline. Your program should perform type checking and raise an appropriate error in the event that either of $\texttt{infile}$ or $\texttt{outfile}$ are not strings. $\texttt{reverse}\_\texttt{lines}$ should open $\texttt{infile}$ for reading and open $\texttt{outfile}$ for writing (and raise an error if either of these operations fail). Your program should overwrite outfile if it already exists.**

***Hint:* you may find it useful to write a function $\texttt{reverse}\_\texttt{string}$ that takes a single string argument and returns that string, reversed**

***Second hint:* be careful about the newlines at the ends of lines! When you read a line from input, what happens to the newline at the end? Read the Python documentation carefully and check your code on simple test cases**

````python
def reverse_string(string):

    if not isinstance(string, (str)):
        raise TypeError(f'string = {string} should be of type str.')
    
    result = ""
    for i in range(len(string)):
        result += string[(len(string) - 1) - i]
    return result

def reverse_lines(infile, outfile):
    
    if not isinstance(infile, (str, )):
        raise TypeError(f'infile = {infile} should be of type str.')
    
    if not isinstance(outfile, (str, )):
        raise TypeError(f'outfile = {outfile} should be of type str.')
    
    line_list = []
    with open(infile, 'r') as f:
        for line in f:
            line_list.append( reverse_string(line).replace('\n', '') )
    
    with open(outfile, 'w') as g:
        g.write('\n'.join(line_list))
````

**There are at least two obvious ways (to me, anyway) of implementing $\texttt{reverse}\_\texttt{lines}$:**

* ***Option 1:* Open $\texttt{infile}$ for reading and $\texttt{outfile}$ for writing. Read the lines of $\texttt{infile}$ one at a time, and save them in the entries of a list, say, $\texttt{line}\_\texttt{list}$. Then, once we have finished reading $\texttt{infile}$, iterate over the entries of $\texttt{line}\_\texttt{list}$, reversing each one and writing it to $\texttt{outfile}$.**

* ***Option 2:* Open $\texttt{infile}$ for reading and $\texttt{outfile}$  for writing. Read the lines of $\texttt{infile}$ one at a time, and each time we read a line, reverse it and write it to outfile immediately.**

**There is at least one good reason to prefer Option 2 over Option 1. What is that reason? Hint: if $\texttt{infile}$ is especially large, what happens to the list $\texttt{line}\_\texttt{list}$?**

**Define two new functions, $\texttt{revlines}\_\texttt{storage}$ and $\texttt{revlines}\_\texttt{direct}$, that have the same signature (i.e., take the same arguments) and perform the same function as $\texttt{reverse}\_\texttt{lines}$. $\texttt{revlines}\_\texttt{storage}$ should follow the design in Option 1 and $\texttt{revlines}\_\texttt{direct}$ should follow the design outlined in Option 2. If you happened to follow one of these designs in your solution for $\texttt{reverse}\_\texttt{lines}$, you are free to copy-paste your code (or, better yet, just have your new function call your already-written function), but make sure you are still defining both $\texttt{revlines}\_\texttt{direct}$ and $\texttt{revlines}\_\texttt{storage}$. There is no need to perform error checking in these two functions.**

````python
def revlines_storage(infile, outfile):
    return reverse_lines(infile, outfile)

def revlines_direct(infile, outfile):

    if not isinstance(infile, (str, )):
        raise TypeError(f'infile = {infile} should be of type str.')
    
    if not isinstance(outfile, (str, )):
        raise TypeError(f'outfile = {outfile} should be of type str.')
    
    with open(infile, 'r') as f, open(outfile, 'w') as g:
        
        is_first_line = True
        for line in f:
            if is_first_line:
                g.write( reverse_string(line).replace('\n', '') )
                is_first_line = False
                
            else:
                g.write('\n')
                g.write( reverse_string(line).replace('\n', '') )    
````

$\texttt{revlines}\_\texttt{direct}$ is a better algorithm to reverse texts in file compared to $\texttt{revlines}\_\texttt{storage}$ for a couple of reasons:

* Memory: $\texttt{revlines}\_\texttt{direct}$ manages memory much better than $\texttt{revlines}\_\texttt{storage}$ as there is no reason to allocate memory to save a seperate list.

* Runtimes: $\texttt{revlines}\_\texttt{direct}$ is faster as both files are open, and a line is taken from infile, reversed, and written into outfile. In comparison, $\texttt{revlines}\_\texttt{storage}$ is slower as we first open infile, take all lines, reverse it, save it in a list iteratively, then open outfile, write each line in the list iteratively.

## 2 Counting Word Bigrams (5 points, spent $\approx$ 25 minutes)

**In your previous homework, you wrote a function for counting character bigrams. Now, let’s write a function for counting word bigrams. That is, for each pair of words, say, cat and dog, we want to count how many times the word “cat” occurred immediately before the word “dog”. We will represent this bigram by a tuple, (’cat’, ’dog’). For our purposes, we will ignore all newlines, punctuation and capitalization in our counting. So, as an example, the fragment of poem,**

*Half a league, half a league,*

*Half a league onward,*

*All in the valley of Death*

*Rode the six hundred.*

**includes the bigrams (’half’, ’a’) and (’a’, ’league’) both three times, the bigram (’league’, ’half’) appears twice, while the bigram (’in’, ’the’) appears only once.**

**Write a function $\texttt{count}\_\texttt{bigrams}\_\texttt{in}\_\texttt{file}$ that takes a filename as its only argument. Your function should read from the given file, and return a dictionary whose keys are bigrams (given in the tuple form above), and values are the counts for those bigrams. Again, your function should ignore punctuation, spaces, newlines and capitalization. The strings in your key tuples should be lower-case. Your function should use a try-catch statement to raise an error with an appropriate message to alert the user in the event that the given file cannot be opened, and a different error in the event that the provided argument isn’t a string at all. Hint: you will find the Python function $\texttt{str.strip()}$, along with the string constants defined in the string documentation (https://docs.python.org/3/library/string.html), useful in removing punctuation.**

 **Hint: be careful to check that your function handles newlines correctly. For example, in the poem above, one of the (’league’, ’half’) bigrams spans a newline, but should be counted nonetheless.**
 
 **Note: be careful that your function does not accidentally count the empty string as a word (this is a common bug if you aren’t careful about splitting the input text). Solutions that merely delete “bad” keys from the dictionary at the end will not receive full credit.**

````python
def remove_punctuations(string):
    
    if not isinstance(string, (str, )):
        raise TypeError(f'string = {string} should be of type str.')

    result = ""
    punctuations = '!"#$%&\'()*+,-./:;<=>?@[\\]^_`{|}~' # derived from running string.punctuations on a seperate cell
    for char in string:
        if char not in punctuations:
            result += char
    return result

def count_bigrams_in_file(filename):
    
    if not isinstance(filename, (str, )):
        raise TypeError(f'filename = {filename} should be of type str.')

    with open(filename, 'r') as f:
        paragraph = ""
        for line in f:
            paragraph += line + " " # write each line into a long continuous paragraph
    
    paragraph = paragraph.lower() # makes all characters lower-case
    paragraph = remove_punctuations(paragraph) # removes all punctuations (defined above)
    paragraph = paragraph.split() # creates a list of all words before and after whitespaces and newlines
    
    dictionary = dict()
    for i in range(len(paragraph) - 1):
        if (paragraph[i], paragraph[i+1]) not in dictionary:
            dictionary[(paragraph[i], paragraph[i+1])] = 1
        else:
            dictionary[(paragraph[i], paragraph[i+1])] += 1
    return dictionary
````

**We say that word A is collocated with word B in a text if words A and B occur immediately one after another (in either order). That is, words A and B are collocated if and only if either of A B or B, A are present in the text.** 

**Write a function $\texttt{collocations}$ that takes a filename as its only argument and returns a dictionary. Your function should read from the given file (raising an appropriate error if the file cannot be opened or if the argument isn’t a string at all) and return a dictionary whose keys are all the strings appearing in the file (again ignoring case and stripping away all spaces, newlines and punctuation) and the value of word A is a Python set object2 containing all the words collocated with A. Again using the poem fragment above as an example, the string ’league’ should appear as a key, and should have as its value the set {’a’, ’half’, ’onward’}, while the string ’in’ should have the set {’all’, ’the’} as its value.** 

**Hint: we didn’t discuss Python sets in lecture, because they are essentially just dictionaries without values. See the documentation for more information**

````python
def collocations(filename):

    if not isinstance(filename, (str, )):
        raise TypeError(f'filename = {filename} should be of type str.')

    with open(filename, 'r') as f:
        paragraph = ""
        for line in f:
            paragraph += line + " " # write each line into a long continuous paragraph
    
    paragraph = paragraph.lower() # makes all characters lower-case
    paragraph = remove_punctuations(paragraph) # removes all punctuations (defined above)
    paragraph = paragraph.split() # creates a list of all words before and after whitespaces and newlines
    
    dictionary = dict()
    for i in range(1, len(paragraph) - 1):
        if paragraph[i] not in dictionary:
            dictionary[paragraph[i]] = set() # initialize empty set
        dictionary[paragraph[i]].add(paragraph[i-1]) # append preceding word
        dictionary[paragraph[i]].add(paragraph[i+1]) # append succeeding word
    return dictionary
````
