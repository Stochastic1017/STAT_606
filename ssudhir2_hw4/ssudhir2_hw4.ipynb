```python
class Vector:
    '''
    A class Vector to represent a vector in n-dimensional space.

    Each instance of the class has two attributes:
    - dim : dimension of the vector.
    - entries : a tuple of n entries of the vector.

    The class has the following methods:
    - __init__ : initializes the vector with dimension and entries.
    - get_entries : returns the entries of the vector.
    - get_dim : returns the dimension of the vector.
    - __str__, __repr__ : returns a string representation of the vector.
    - __compare__, __eq__, __lt__, __le__, __gt__, __ge__ : component-wise comparison operators.
    - __operation__, __add__, __sub__ : addition and subtraction operators.
    - dot : dot product of two vectors.
    - __mul__, __rmul__ : scalar multiplication operators.
    - norm : p-norm of the vector.

    Example usage:
    >>> v1 = Vector(3, [1, 2, 3])
    >>> v2 = Vector(3, (4, 5, 6))
    >>> v1 + v2
    Vector(3, (5, 7, 9))
    '''
    def __init__(self, dim, entries = None):

        if not isinstance(dim, (int, )):
            raise TypeError(f'dimension {dim} should be of type int.') # raises error if input is not an integer
        if dim <= 0:
            raise ValueError(f'dimension {dim} should be a positive integer.') # raises error if input is not positive
        
        if entries is None:
            entries = (0, ) * dim # creates a tuple of dim zeros
        elif isinstance(entries, (list)):
            entries = tuple(entries) # converts list to tuple
        else:
            if not isinstance(entries, (tuple, list, )):
                raise TypeError(f'entries {entries} should be of type tuple or list.') # raises error if input is not a tuple or list
                
        if any(not isinstance(item, (int, float)) for item in entries):
            raise ValueError(f'entries {entries} should have entries int or float only.') # raises error if input is not an integer or float

        if len(entries) != dim:
            raise ValueError(f'entries {entries} should match dimension dim = {dim}.') # raises error if input does not match dimension

        self.dim = dim
        self.entries = entries

    def __str__(self): # returns a string representation of the vector using __str__()
        return f'Vector is of dimension {self.get_dim()} and has entries {self.get_entries()}'

    def __repr__(self): # returns a string representation of the vector using __repr__()
        return f'Vector({self.get_dim()}, {self.get_entries()})'

    def get_dim(self): # returns the dimension of the vector
        return self.dim
    
    def get_entries(self): # returns the entries of the vector
        return self.entries
    
    ################# Equality and Comparison Operators #################

    def __compare__(self, other): # placeholder function to check <, >, and ==

        if not isinstance(other, (Vector, )):
            raise TypeError(f'both vectors should be of type Vector.') # raises error if input is not of class Vector

        if self.get_dim() != other.get_dim():
            raise ValueError(f'both vectors {self.get_entries()} and {other.get_entries()} should have the same dimensions.') # raises error if dimensions do not match
        
        for index in range(self.dim):
            if self.get_entries()[index] < other.get_entries()[index]: # checks if each entry of self is less than other
                return -1
            elif self.get_entries()[index] > other.get_entries()[index]: # checks if each entry of self is greater than other
                return 1
        return 0

    def __eq__(self, other): # strictly equal to (==)
        return self.__compare__(other) == 0
    
    def __lt__(self, other): # strictly lesser than (<)
        return self.__compare__(other) == -1 
    
    def __le__(self, other): # less than OR equal to (<=)
        return (self.__compare__(other) == -1 or self.__compare__(other) == 0)
    
    def __gt__(self, other): # strictly greater than (>)
        return self.__compare__(other) == 1
    
    def __ge__(self, other): # greater than OR equal to (>=)
        return (self.__compare__(other) == 1 or self.__compare__(other) == 0)
    
    ################# Addition and Subtraction Operators #################

    def __operation__(self, other, operator): # placeholder function to check + and -

        if not isinstance(other, (Vector, )):
            raise TypeError(f'both vectors should be of type Vector.') # raises error if input is not of class Vector

        if self.get_dim() != other.get_dim():
            raise ValueError(f'both vectors {self.get_entries()} and {other.get_entries()} should have the same dimensions.') # raises error if dimensions do not match
        
        result = ()
        if operator == '+':           
            for i in range(self.get_dim()):
                result += (self.get_entries()[i] + other.get_entries()[i], ) # adds each entry of self and other
        elif operator == '-':
            for i in range(self.get_dim()):
                result += (self.get_entries()[i] - other.get_entries()[i], ) # subtracts each entry of self and other
        else:
            raise SyntaxError(f'{operator} is an invalid operator.') # raises error if operator is invalid
        return Vector(len(result), result)

    def __add__(self, other): # adding to Vectors (+)
        return self.__operation__(other, '+')
        
    def __sub__(self, other): # subtracting Vectors (-)
        return self.__operation__(other, '-')

    ###################### Dot Product Operator #######################

    def dot(self, other):
         
        if not isinstance(other, (Vector, )):
            raise TypeError(f'both vectors should be of type Vector.') # raises error if input is not of class Vector

        if self.get_dim() != other.get_dim():
            raise ValueError(f'both vectors {self.get_entries()} and {other.get_entries()} should have the same dimensions.') # raises error if dimensions do not match
        
        result = 0
        for i in range(self.get_dim()):
            result += (self.get_entries()[i] * other.get_entries()[i]) # multiplies each entry of self and other and adds them
        return result
    
    #################### Scalar Multiply Operator ######################

    def __mul__(self, other):

        if not isinstance(other, (Vector, int, float, )):
            raise TypeError(f'{other} should be of type Vector, int or float.') # raises error if input is not of class Vector, int, or float
        
        if isinstance(other, (Vector, )): # if other is a Vector, i.e., vector multiplication
            if self.get_dim() != other.get_dim():
                raise ValueError(f'both vectors {self.get_entries()} and {other.get_entries()} should have the same dimensions.') # raises error if dimensions do not match
                                   
            result = ()
            for i in range(self.get_dim()):
                result += (self.get_entries()[i] * other.get_entries()[i], ) # multiplies each entry of self and each entry of other
            return Vector(len(result), result)
        
        else: # if other is an int or float, i.e., scalar multiplication
            result = ()
            for i in range(self.get_dim()):
                result += (self.get_entries()[i] * other, ) # multiplies each entry of self with other
            return Vector(len(result), result)
        
    def __rmul__(self, other):

        if isinstance(other, (Vector, )):
            return other.__mul__(self) # calls __mul__ method of other
        else:
            return self.__mul__(other) # calls __mul__ method of self
        
    ################## p-norm operation ####################
        
    def norm(self, p = 2): # by default p = 2 euclidean norm calculated
        
        if p in ('inf', 'Infinity', 'infinity', ): 
            p = float('inf') 
        if not isinstance(p, (int, float, )):
            raise TypeError(f'norm {p} must be of type int, float, or "inf" / "Infinity".') # raises error if p is not of type int, float, or "inf" / "Infinity"
        if p < 0:
            raise ValueError(f'norm {p} must be non-negative.') # raises error if p is negative
        
        if p == 0:
            return sum(1 for entry in self.get_entries() if entry != 0) # counts the number of non-zero entries (for p = 0)
        
        if p == float('inf'):
            return max(self.get_entries()) # calculates infinity norm (for p = ∞)
        
        else:
            return (sum(entry**p for entry in self.get_entries()))**(1/p) # calculates p-norm (for 0 < p < ∞)
```

**Did you choose to make the vector’s entries a tuple or a list (there is no strictly right or wrong answer here, although I would say one is better than the other in this context)? Defend your choice.**

I made my vector's entries a `tuple` instead of a `list` for the following couple of reasons:

* `Vector.get_dim()` requires the output to be a tuple. Due to this, we are saving the run-time that would be required to convert a `list` to a `tuple` for the `get_dim()` method.

* `tuples` are immutable objects, and therefore all the methods, operators, and comparisons defined with input as a `tuple` object can be greatly controlled, and other confounding mutations do not need to be taken into account.

* `tuples` are also inherently faster than lists and have direct access to their pointers, thus making them an ideal choice in this scenario.

**Are the `dim` and `entries` class attributes or instance attributes? Why is this the right design choice?**

In the Vector class, both `dim` and `entries` are **instance** attributes. We know this because for every unique `Vector` object, we will have unique `dim` and `entries` for each object. If instead these were **class** attributes, then the user would not have any agency in choosing the `dim` and `entries`, and every `Vector` object would effectuvely be the same. Therefore, we can conclude that `dim` and `entries` being instance attributes is the right design choice.

## 2.) Random Walkers (7 points, spent $\approx$ 20 minutes)

**In this problem, you’ll get some practice working with inheritance by building a collection of objects for generating and representing random walks. Recall that a random walk is a discrete-time stochastic process, in which a “walker” starts out at some position $S_0$ at time $0$, and for all $t = 1,2, \dots$, the walker’s position at time $t + 1$ is given by $S_{t+1} = S_t + X_t$, where $X_1, X_2, \dots$ are drawn i.i.d. from some distribution. The variables $X_1, X_2, \dots$ are called the “steps” or “increments” of the random walk.**

**The simplest random walk is the symmetric random walk on the integers. The walker starts out at $0$ at time $t = 0$, and the steps $X_1, X_2, \dots$ are independent Rademacher random variables. That is, for all $i = 1, 2, \dots$, we have that $\mathbb{P}[X_i = 1] = \mathbb{P}[X_i = −1] = 1/2$. At each time $t = 0, 1, 2, \dots$ , we say that the current position of the random walker is $S_t = \sum_{i=1}^{t} X_i$, so that $S_0 = 0$, $S_1 = X_1$, $S_2 = X_1 + X_2$, and so on. At any given time $t$, we can speak of the “history” or the “sample path” of the random walker, $(S_0, S_1, . . . , S_{t−1})$. That is, the sequence of integers that the random walker has visited prior to time $t$.**

**Implement a class representing symmetric random walk on the integers, called `SymIntRW`, which has the following instance attributes:**

* **`current_position`: an integer, the current position of the random walker.**

* **`history`: a list of integers that specify the previous positions of the walker, in order (so that the last entry of the list is the most recent previous position of the walker).**

**The class should support the following methods:**
* **`get_time()`: return a non-negative integer encoding the number of steps that the walker has taken so far. That is, return the length of the history attribute.**

* **`get_history()`: return the history attribute, i.e., the list of previous positions.**

* **`get_position()`: return the current_position attribute, i.e., the integer that the walker is currently at.**

* **`step()`: take a single step. Generate a random step according to a Rademacher distribution. Update the current position and history accordingly. Hint: use the Python `random.randint` function.**

**You should also implement an initialization method that takes no arguments (aside from the self argument) and initializes `self.current_position` to $0$ and `self.history` to be the empty list. This reflects the fact that initially, the random walker starts at $S_0 = 0$, at which point it has no “history”.**

```python
import random 

class SymIntRW:
    '''
    A class to represent a symmetric integer random walk.

    Each instance of the class has two attributes:
    - current_position : the current position of the random walk.
    - history : a list to store the history of positions.

    The class has the following methods:
    - __init__ : initializes the random walk with current position and history.
    - step : takes a step in the random walk.
    - get_time : returns the number of steps taken.
    - get_history : returns the history of positions.
    - get_position : returns the current position.

    Example usage:
    >>> rw = SymIntRW()
    >>> rw.step()
    >>> rw.step()
    >>> rw.step()
    >>> print(rw.get_position())
    >>> print(rw.get_time())
    >>> print(rw.get_history())
    '''
    def __init__(self):
        self.current_position = 0 # initial position
        self.history = [] # list to store the history of positions

    def step(self):
        forward = random.choice([-1, 1]) # randomly choose to move forward or backward with 1/2 probability
        self.current_position += forward # update the current position
        self.history.append(self.current_position) # append the current position to the history
    
    def get_time(self):
        return len(self.history) # returns the number of steps taken
    
    def get_history(self):
        return self.history # returns the history of positions
    
    def get_position(self):
        return self.current_position # returns the current position
    
```

```python
RW = SymIntRW()
number_of_steps = 5
print('Initial position', RW.get_position())
print('')
for i in range(number_of_steps):
    RW.step()
    print('Next position', RW.get_position())
    print('History', RW.get_history())
    print('Time', RW.get_time())
    print('')
```

**One natural extension of the symmetric random walk on the integers is to allow the steps to be drawn from a different distribution. Implement a class `IntegerRW` that inherits from `SymIntRW`, but has the additional attribute `p`, where `p` is a probability. Override the following methods:**

* **Override the initialization method to take a single optional argument `p`, which should default to `0.5`, which initializes the `self.p` attribute to be equal to the argument `p`, and raises an appropriate error in the event that `p` is not a probability (i.e., is not an int or float in the range [0, 1]).**

* **Override the step method so that the steps are generated according to $\mathbb{P}[X_i = 1] = 1 − \mathbb{P}[X_i = −1] = p$. Hint: `random.random()` will generate a random float drawn uniformly from the interval $[0, 1)$.**

```python
class IntegerRW( SymIntRW ):
    '''
    A class to represent a symmetric integer random walk, with probability p of moving forward.

    Each instance of the class inherits two attributes from the parent class:
    - current_position : the current position of the random walk.
    - history : a list to store the history of positions.

    Each instance of the class has one additional attribute:
    - p : probability of moving forward.

    The class inherits the following methods:
    - __init__ : initializes the random walk with current position and history.
    - step : takes a step in the random walk.
    - get_time : returns the number of steps taken.
    - get_history : returns the history of positions.
    - get_position : returns the current position.

    Example usage:
    >>> rw = IntegerRW(p = 0.9)
    >>> rw.step() 
    >>> rw.step()
    >>> rw.step()
    >>> print(rw.get_position())
    >>> print(rw.get_time())
    >>> print(rw.get_history())
    '''
    def __init__(self, p = 0.5): # default probability of moving forward is 0.5
        super().__init__() # calls the __init__ method of the parent class SymIntRW
        
        if not isinstance(p, (int, float, )):
            raise TypeError(f'{p} should be of type int or float.') # raises error if input is not an integer or float
        if p < 0 or p > 1:
            raise ValueError(f'{p} should be between 0 and 1.') # raises error if input is not between 0 and 1
        
        self.p = p # probability of moving forward

    def step(self):

        threshold = random.random() # generates a random number uniformly between 0 and 1

        if self.p > threshold:
            forward = 1 # move forward with probability p
        else:
            forward = -1 # move backward with probability 1 - p
        
        self.current_position += forward # update the current position
        self.history.append(self.current_position) # append the current position to the history
```

**Another extension of the symmetric random walk on the integers is to define the random walk on 2-, 3- or higher-dimensional space. The d-dimensional simple symmetric random walk (SSRW) is a random walk on the integer lattice**

$$\mathbb{Z}^d = \{(z_1, \dots, z_d) : \forall z_i \in \mathbb{Z}\}$$

**in which the steps $X_1, X_2, \dots \in \mathbb{Z}^d$ are drawn i.i.d. from the uniform distribution on**

$$ \{z \in \mathbb{z}^d : ||z|| = 1\} $$

**Said another way, to generate a step $X_t$, we generate $X_t = \epsilon_t Z_t$, where $\epsilon_t \in \{−1, 1\}$ is a Rademacher random variable,**

$$\mathbb{P}[\epsilon_t = 1] = \mathbb{P}[\epsilon_t = -1] = 1/2$$

**and $Z_t$ is generated by choosing an index $j$ uniformly from the set of integers $\{1, 2, \dots , d\}$, and setting**

$$(Z_{t})_i = 
\begin{cases}
1 & \text{if}\;i = j\\
0 & \text{if}\;i \neq j
\end{cases}
$$

**This process is called the symmetric random walk because the steps are chosen uniformly from the points adjacent to the origin in $\mathbb{Z}^d$ (i.e., so that the step distribution is symmetric about the origin).**

```python
class SSRW( SymIntRW ):
    '''
    A class to represent a symmetric integer random walk, with d-dimensional of vectors.

    Each instance of the class inherits two attributes from the parent class:
    - `current_position` -> the current position of the random walk.
    - `history` -> a list to store the history of positions.

    Each instance of the class has one additional attribute:
    - `dimension` -> dimension of the vector.

    The class inherits the following methods:
    - `__init__` -> initializes the random walk with current position and history.
    - `step` -> takes a step in the random walk.
    - `get_time` -> returns the number of steps taken.
    - `get_history` -> returns the history of positions.
    - `get_position` -> returns the current position.

    Example usage:
    >>> rw = SSRW(d = 5)
    >>> rw.step() 
    >>> rw.step()
    >>> rw.step()
    >>> print(rw.get_position())
    >>> print(rw.get_time())
    >>> print(rw.get_history())
    '''
    def __init__(self, d):
        super().__init__() # calls the __init__ method of the parent class SymIntRW

        if not isinstance(d, (int, )):
            raise TypeError(f'dimension {d} must be of type int.') # raises error if input is not an integer
        if d <= 0:
            raise ValueError(f'dimension {d} must be positive.') # raises error if input is not positive
        
        self.dimension = d # dimension of the vector
        self.current_position = [0] * self.dimension # initial position in d-dimensional space

    def step(self):
        
        forward = random.choice([-1, 1]) # randomly choose to move forward or backward with 1/2 probability
        j = random.randint(0, self.dimension - 1) # randomly choose a direction to move

        self.current_position[j] += forward # update the current position
        self.history.append(self.current_position.copy()) # append the current position to the history
```

```python
RW = SSRW(d = 5)
number_of_steps = 5
print('Initial position', RW.get_position())
print('')
for i in range(number_of_steps):
    RW.step()
    print('Next position', RW.get_position())
    print('History', RW.get_history())
    print('Time', RW.get_time())
    print('')
```
